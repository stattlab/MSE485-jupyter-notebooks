# Useful Utility: String Replace and Regular Expression

Regular expressions are extremely powerful: https://docs.python.org/3/howto/regex.html 

```python
string = "data/T_1.0/p_0.5/test.gsd" 
new_string_replace = string.replace("test","init") 
print(new_string_replace)

matches = re.findall(r"[+-]? *(?:\d+(?:\.\d*)?|\.\d+)(?:[eE][+-]?\d+)?", string)
print(matches)
```

# GLOB to traverse folders and files

```python
match_floats = re.compile(r"[+-]? *(?:\d+(?:\.\d*)?|\.\d+)(?:[eE][+-]?\d+)?")

files = glob.glob("./data/*/*/equi.gsd") 

for file in files: 
    parameters = match_floats.findall(file)
    print(file,parameters)
```

```python
for file in files: 
    if 'T_3.0' in file:
        print(file)
```

Without glob, if file-structure is perfectly regular and organized:

```python
for T in [1,2,3]: 
    for p in [0,0.5,1.5]: 
        filename = "./data/T_%1.1f_p_%1.1f/run.gsd"%(T,p) 
        print(filename)
```

```python
# glob.glob with the ** pattern and recursive=True
# match all files and directories within the data directory and its subdirectories
all_files_and_dirs = glob.glob('./data/**', recursive=True)

# Filter out only the files (excluding directories)
all_files = [f for f in all_files_and_dirs if os.path.isfile(f)]

# Print the list of all files found
for file_path in all_files:
    print(file_path)
```
